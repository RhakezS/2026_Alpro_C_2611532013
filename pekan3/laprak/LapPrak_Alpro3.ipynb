## 2.1 Dasar Teori
Python memiliki berbagai jenis operator yang digunakan untuk melakukan proses tertentu terhadap data dan variabel. Operator merupakan simbol atau kata khusus yang digunakan untuk melakukan operasi, seperti perhitungan, perbandingan, pemberian nilai, maupun pengolahan kondisi. Penggunaan operator menjadi salah satu dasar penting dalam pemrograman karena hampir setiap program membutuhkan operasi terhadap data.

1. Operator Aritmatika
Operator aritmatika digunakan untuk melakukan operasi perhitungan matematika pada Python. Operator yang termasuk di dalamnya adalah penjumlahan (+), pengurangan (-), perkalian (*), pembagian (/), pembagian bulat (//), modulus atau sisa pembagian (%), dan perpangkatan (**). Operator aritmatika dapat digunakan pada berbagai tipe data numerik, seperti integer dan float. Penggunaannya banyak ditemukan dalam perhitungan nilai, transaksi, luas bangun, dan berbagai proses perhitungan lainnya.
2. Operator Perbandingan
Operator perbandingan digunakan untuk membandingkan dua nilai atau lebih. Hasil dari operasi perbandingan berupa nilai Boolean, yaitu True atau False. Operator yang digunakan meliputi sama dengan (==), tidak sama dengan (!=), lebih besar (>), lebih kecil (<), lebih besar atau sama dengan (>=), serta lebih kecil atau sama dengan (<=). Operator ini biasanya digunakan dalam percabangan dan pemeriksaan kondisi pada suatu program.
3. Operator Logika
Operator logika digunakan untuk menggabungkan atau memanipulasi beberapa kondisi. Python memiliki tiga operator logika utama, yaitu and, or, dan not. Operator and menghasilkan nilai benar apabila semua kondisi yang diperiksa bernilai benar, sedangkan or menghasilkan nilai benar apabila terdapat setidaknya satu kondisi yang benar. Sementara itu, not digunakan untuk membalik nilai Boolean dari suatu kondisi. Operator logika banyak digunakan dalam proses pengambilan keputusan pada program.
4. Operator Penugasan
Operator penugasan digunakan untuk memberikan nilai kepada variabel atau mengubah nilai yang telah tersimpan di dalam variabel. Operator yang paling dasar adalah =, sedangkan operator gabungan lainnya antara lain +=, -=, *=, /=, //=, %= dan **=. Operator tersebut dapat membuat penulisan program menjadi lebih sederhana, terutama ketika suatu variabel perlu diperbarui berdasarkan nilai sebelumnya.
5. Operator Bitwise
Operator bitwise digunakan untuk melakukan operasi pada tingkat bit dari suatu bilangan integer. Python menyediakan beberapa operator bitwise, yaitu AND (&), OR (|), XOR (^), NOT (~), left shift (<<), dan right shift (>>). Operasi ini bekerja dengan memproses representasi biner dari suatu bilangan. Operator bitwise biasanya digunakan dalam pengolahan data tingkat rendah, manipulasi bit, sistem komputer, dan beberapa kebutuhan pemrograman tertentu.
6. Operator Keanggotaan
Operator keanggotaan digunakan untuk memeriksa apakah suatu nilai terdapat di dalam sebuah objek seperti string, list, tuple, atau set. Python memiliki operator in dan not in. Operator in menghasilkan True apabila suatu nilai ditemukan di dalam objek, sedangkan not in menghasilkan True apabila nilai tersebut tidak ditemukan. Operator ini dapat digunakan untuk melakukan pemeriksaan data dengan cara yang sederhana.
7. Operator Identitas
Operator identitas digunakan untuk mengetahui apakah dua variabel mengacu pada objek yang sama di dalam memori. Python menyediakan operator is dan is not. Operator is menghasilkan True apabila kedua variabel mengacu pada objek yang sama, sedangkan is not menghasilkan True apabila keduanya tidak mengacu pada objek yang sama. Operator identitas berbeda dengan operator == yang digunakan untuk membandingkan nilai.

```python
# Buat file dengan nama aritmatika_NIM.py
# Buat program untuk operator aritmatika dalam Python
# Nama variabel ditambah 4 digit nim terakhir contoh: angka1_1234
# Program ini menngunakan fungsi input()
# Nilai yang dimasukkan akan dikonversi menjadi tipe data integer

angka1_2013 = int(input("Input angka-1: "))
angka2_2013 = int(input("Input angka-2: "))

# Penjumlahan
hasil_2013 = angka1_2013 + angka2_2013
print("\nOperator Penjumlahan")
print("Hasil =",hasil_2013)

# Pengurangan
hasil_2013 = angka1_2013 - angka2_2013
print("\nOperator Pengurangan")
print("Hasil =",hasil_2013)

# Perkalian
hasil_2013 = angka1_2013 * angka2_2013
print("\nOperator Perkalian")
print("Hasil =",hasil_2013)

# Pembagian, pembagian bulat, dan sisa bagi
if angka2_2013 != 0:
    hasil_2013 = angka1_2013 / angka2_2013
    print("\nOperator Pembagian")
    print("Hasil =",hasil_2013)

    hasil_2013 = angka1_2013 // angka2_2013
    print("\nOperator Pembagian Bulat")
    print("Hasil =",hasil_2013)

    hasil_2013 = angka1_2013 % angka2_2013
    print("\nOperator Sisa Bagi")
    print("Hasil =",hasil_2013)
else:
    print("Angka kedua tidak boleh bernilai 0.")

# Pangkat
hasil_2013 = angka1_2013 ** angka2_2013
print("\nOperator Pangkat")
print("Hasil =",hasil_2013)
```

```python
# Buat file dengan nama assigment_NIM.py
# Nama variabel ditambah 4 digit nim terakhir contoh: angka1_1234
# Program ini menngunakan fungsi input()
# Nilai yang dimasukkan akan dikonversi menjadi tipe data integer
# Program operator assigment dalam Python

angka1_2013 = int(input("Input angka-1: "))
angka2_2013 = int(input("Input angka-2: "))

print("\nNilai awal angka1 =",angka1_2013)
print("Nilai angka2 =",angka2_2013)

# Assigment biasa
hasil_2013 = angka1_2013 
print("\nAssigment biasa (=)")
print("Hasil =",hasil_2013)

# Assigment penambahan
hasil_2013 = angka1_2013 
hasil_2013 += angka2_2013 
print("\nAssigment penambahan (+=)")
print("Hasil =",hasil_2013)

# Assigment pengurangan
hasil_2013 = angka1_2013 
hasil_2013 -= angka2_2013 
print("\nAssigment pengurangan (-=)")
print("Hasil =",hasil_2013)

# Assigment perkalian
hasil_2013 = angka1_2013 
hasil_2013 *= angka2_2013
print("\nAssigment perkalian (*=)")
print("Hasil =",hasil_2013)

# Assigment pembagian, pembagian bulat, dan sisa bagi
if angka2_2013 != 0:
    hasil_2013 = angka1_2013 
    hasil_2013 /= angka2_2013
    print("\nAssigment pembagian (/=)")
    print("Hasil =",hasil_2013)
    # Operator tambahan
    hasil_2013 = angka1_2013 
    hasil_2013 //= angka2_2013
    print("\nAssigment pembagian bulat (//=)")
    print("Hasil =",hasil_2013)
    hasil_2013 = angka1_2013 
    hasil_2013 %= angka2_2013
    print("\nOperator sisa bagi (%=)")
    print("Hasil =",hasil_2013)
else:
    print("\nPembagian tidak dapat dilakukan.")
    print("Angka kedua tidak boleh bernilai 0.")

# Operator tambahan: assigment perpangkatan
hasil_2013 = angka1_2013 
hasil_2013 **= angka2_2013
print("\nOperator perpangkatan (**=)")
print("Hasil =",hasil_2013)
```

```python
# Buat file dengan nama perbandingan_NIM.py
# Nama variabel ditambah 4 digit nim terakhir contoh: angka1_1234
# Program ini menngunakan fungsi input()
# Nilai yang dimasukkan akan dikonversi menjadi tipe data integer
# Program operator perbandingan dalam Python

angka1_2013 = int(input("Input angka-1:"))
angka2_2013 = int(input("Input angka-2:"))

# Lebih besar dari
hasil_2013 = angka1_2013 > angka2_2013
print("\nOperator lebih besar dari")
print("angka1_2013 > angka2_2013 =", hasil_2013)

# Lebih kecil dari
hasil_2013 = angka1_2013 < angka2_2013
print("\nOperator lebih kecil dari")
print("angka1_2013 < angka2_2013 =", hasil_2013)

# Lebih besar dari atau sama dengan
hasil_2013 = angka1_2013 >= angka2_2013
print("\nOperator lebih besar dari atau sama dengan")
print("angka1_2013 >= angka2_2013 =", hasil_2013)

# Lebih kecil dari atau sama dengan
hasil_2013 = angka1_2013 <= angka2_2013
print("\nOperator lebih kecil dari atau sama dengan")
print("angka1_2013 <= angka2_2013 =", hasil_2013)

# Sama dengan
hasil_2013 = angka1_2013 == angka2_2013
print("\nOperator sama dengan")
print("angka1_2013 == angka2_2013 =", hasil_2013)

# Tidak sama dengan
hasil_2013 = angka1_2013 != angka2_2013
print("\nOperator tidak sama dengan")
print("angka1_2013 != angka2_2013 =", hasil_2013)

# Tambahan: Perbandingan berantai dalam python
hasil_2013 = 0 < angka1_2013 < 100
print("\nPerbandingan berantai")
print("0 < angka1_2013 < 100 =", hasil_2013)

hasil_2013 = 0 < angka2_2013 < 100
print("0 < angka2_2013 < 100 =", hasil_2013)
```

### 2.2.4 Program logika_2611532013

```python
# Buat file dengan nama logika_NIM.py
# Nama variabel ditambah 4 digit nim terakhir contoh: a1_1234
# Program ini menngunakan fungsi input()
# Program operator logika dalam Python

# Memasukan nilai boolean
# Input tidak peka terhadap huruf besar dan kecil
a1_2013 = input("Input nilai boolean-1 (true/false): ").strip().lower() == "true"
a2_2013 = input("Input nilai boolean-2 (true/false): ").strip().lower() == "true"

print("\nA1 = ",a1_2013)
print("A2 = ",a2_2013)

# Konjungsi: bernilai True jika keduanya True
hasil_2013 = a1_2013 and a2_2013
print("\nKonjungsi (AND)")
print("A1 and A2 =",hasil_2013)

# Disjungsi: bernilai True jika salah satunya False
hasil_2013 = a1_2013 or a2_2013
print("\nDisjungsi (OR)")
print("A1 or A2 =",hasil_2013)

# Negasi A1: membalik nilai A1
hasil_2013 = not a1_2013
print("\nNegasi A1(NOT)")
print("not A1",hasil_2013)

# Negasi A2: membalik nilai A2
hasil_2013 = not a2_2013
print("\nNegasi A2(NOT)")
print("not A2",hasil_2013)

# XOR: bernilai True jika kedua nilai berbeda
hasil_2013 = a1_2013 != a2_2013
print("\nDisjungsi Ekslusif (XOR)")
print("A1 XOR A2",hasil_2013)
```

<div style="font-family: 'Times New Roman', Times, serif;font-size: 18px;">

Penjelasan:
1. Memasukkan nilai Boolean<br>
a1_2013 = input("Input nilai boolean-1 (true/false): ").strip().lower() == "true"<br>
a2_2013 = input("Input nilai boolean-2 (true/false): ").strip().lower() == "true"<br>
Program meminta pengguna memasukkan nilai true atau false.<br>
input() menerima masukan dari pengguna.<br>

```python
# Buat file dengan nama bitwise_NIM.py
# Nama variabel ditambah 4 digit nim terakhir contoh: angka1_1234
# Program ini menggunakan fungsi input()

print("======================================")
print("3. OPERATOR BITWISE")
print("======================================")

angka1_2013 = int(input("Masukkan angka bitwise-1: "))
angka2_2013 = int(input("Masukkan angka bitwise-2: "))

print("\nAngka dalam bentuk desimal dan biner")
print("angka1 =",angka1_2013,"| biner =",bin(angka1_2013))
print("angka1 =",angka2_2013,"| biner =",bin(angka2_2013))

# Bitwise AND
hasil_2013 = angka1_2013 & angka2_2013
print("\nBitwise AND (&)")
print(angka1_2013,"&",angka2_2013,hasil_2013)
print("Biner hasil =",bin(hasil_2013))
print("Biner hasil (8 bit) =",format(hasil_2013,"08b"))

# Bitwise OR
hasil_2013 = angka1_2013 | angka2_2013
print("\nBitwise OR (|)")
print(angka1_2013,"|",angka2_2013,hasil_2013)
print("Biner hasil =",bin(hasil_2013))
print("Biner hasil (8 bit) =",format(hasil_2013,"08b"))

# Bitwise XOR
hasil_2013 = angka1_2013 ^ angka2_2013
print("\nBitwise XOR (^)")
print(angka1_2013,"^",angka2_2013,hasil_2013)
print("Biner hasil =",bin(hasil_2013))
print("Biner hasil (8 bit) =",format(hasil_2013,"08b"))

# Bitwise NOT
hasil_2013 = ~angka1_2013
print("\nBitwise NOT (~)")
print(angka1_2013,"~",angka2_2013,hasil_2013)
print("Biner hasil =",bin(hasil_2013))
print("Biner hasil (8 bit) =",format(hasil_2013,"08b"))

# Bitwise geser kiri
jumlah_geser_2013 = int(input("\nMasukkan jumlah pergeseran bit: "))

hasil_2013 = angka1_2013 << jumlah_geser_2013
print("\nBitwise geser kiri (<<)")
print(angka1_2013,"<<",jumlah_geser_2013,"=",hasil_2013)
print("Biner hasil =",bin(hasil_2013))
print("Biner hasil (8 bit) =",format(hasil_2013,"08b"))

# Bitwise geser kanan
hasil_2013 = angka1_2013 >> jumlah_geser_2013
print("\nBitwise geser kanan (>>)")
print(angka1_2013,">>",jumlah_geser_2013,"=",hasil_2013)
print("Biner hasil =",bin(hasil_2013))
print("Biner hasil (8 bit) =",format(hasil_2013,"08b"))
```

print(f"Diskon : Rp{diskon_2013 * total_2013}")<br>
Menghitung nilai diskon dengan mengalikan total belanja dengan 5%.<br>
Contoh jika belanja Rp200.000:<br>
0.05 × 200000 = Rp10.000
14. Menghitung total pembayaran<br>
print(f"Total Pembayaran : Rp{total_2013 - total_2013 * diskon_2013}")<br>
Mengurangi total belanja dengan jumlah diskon untuk mendapatkan total yang harus dibayar.
15. Menghitung rata-rata harga barang<br>
print(f"Rata-rata Harga Barang : Rp{(total_2013 - total_2013 * diskon_2013)/jumlah_2013}")<br>
Menghitung rata-rata harga setiap barang setelah diskon dengan membagi total pembayaran dengan jumlah barang.
16. Menampilkan hak akses pelanggan<br>
print(f"Member Access : {status_valid_2013}")<br>
print(f"Promo Access : {promo_valid_2013}")<br>
Menampilkan hak akses berdasarkan status member dan validitas kode promo.<br>
Jika pelanggan adalah member, Member Access bernilai True. Jika kode promo tersedia, Promo Access bernilai True.
17. Membuat kode transaksi<br>
kode_transaksi_2013 = int(status_valid_2013) << 0 | int(syarat_total_2013) << 1 | int(syarat_jumlah_2013) << 2 | int<br>(promo_valid_2013) << 3
Bagian ini mengubah nilai Boolean menjadi angka dan menyusunnya menggunakan operasi bitwise.<br>
True dikonversi menjadi 1, sedangkan False menjadi 0.<br>
Setiap kondisi ditempatkan pada posisi bit berbeda:<br>
Status member → bit 0<br>
Total belanja → bit 1<br>
Jumlah barang → bit 2<br>
Promo → bit 3<br>
Operator << digunakan untuk menggeser bit, sedangkan | digunakan untuk menggabungkan semua bit tersebut.
18. Membuat kode referensi<br>
kode_referensi_2013 = int(status_valid_2013) << 0 | int(syarat_total_2013) << 1 | int(promo_valid_2013) << 3<br>
Membuat kode referensi dengan menggunakan status member, total belanja, dan promo. Kondisi jumlah barang tidak dimasukkan ke dalam kode ini.
19. Menampilkan kode biner setiap kondisi<br>
print(f"{format(int(status_valid_2013) << 0,"04b")} | {format(int(syarat_total_2013) << 1,"04b")} | {format(int<br>(syarat_jumlah_2013) << 2,"04b")} | {format(int(promo_valid_2013) << 3,"04b")}")
Menampilkan masing-masing kondisi dalam bentuk biner 4 bit.<br>
Fungsi:<br>
format(nilai, "04b")<br>
digunakan untuk mengubah nilai menjadi biner dengan panjang minimal 4 bit.
20. Menampilkan kode transaksi<br>
print(f"Kode Biner   : {format(kode_transaksi_2013,"04b")}")<br>
print(f"Kode Desimal : {kode_transaksi_2013}")<br>
Menampilkan kode transaksi dalam bentuk biner dan desimal.
21. Pemeriksaan status member<br>
print(f"{format(kode_transaksi_2013,"04b")} & {format(int(status_valid_2013) << 0,"04b")}")<br>
Menampilkan operasi bitwise AND antara kode transaksi dan bit status member.<br>
print(f"Hasil Biner   : {format((kode_transaksi_2013) & int(status_valid_2013) << 0,"04b")}")<br>
print(f"Hasil Desimal : {(kode_transaksi_2013) & int(status_valid_2013) << 0}")<br>
Menghasilkan nilai berdasarkan pemeriksaan bit status member.
22. Pemeriksaan promo<br>
print(f"{format(kode_transaksi_2013,"04b")} & {format(int(promo_valid_2013) << 3,"04b")}")<br>
Menampilkan operasi AND untuk memeriksa bit promo.<br>
print(f"Hasil Biner   : {format((kode_transaksi_2013) & int(promo_valid_2013) << 3,"04b")}")<br>
print(f"Hasil Desimal : {(kode_transaksi_2013) & int(promo_valid_2013) << 3}")<br>
Menampilkan hasil pemeriksaan promo dalam bentuk biner dan desimal.
23. Menampilkan kode transaksi dan referensi<br>
print(f"Kode Transaksi : {format(kode_transaksi_2013,"04b")}")<br>
print(f"Kode Referensi : {format(kode_referensi_2013,"04b")}")<br>
Menampilkan kedua kode dalam bentuk biner 4 bit untuk dibandingkan.
24. Membandingkan kode dengan XOR<br>
print(f"{format(kode_transaksi_2013,"04b")} ^ {format(kode_referensi_2013,"04b")}")<br>
Operator ^ digunakan untuk membandingkan bit dari kode transaksi dan kode referensi. Bit yang berbeda akan menghasilkan 1.<br>
print(f"Hasil Biner   : {format((kode_transaksi_2013) ^ (kode_referensi_2013),"04b")}")<br>
Baris ini menghitung XOR antara kode_transaksi_2013 dan kode_referensi_2013
25. Pergeseran kode transaksi<br>
print(f"{format(kode_transaksi_2013,"04b")} << 1")<br>
Menampilkan operasi pergeseran bit kode transaksi sebanyak 1 posisi ke kiri.<br>
print(f"Hasil Biner   : {format((kode_transaksi_2013) << 1,"04b")}")<br>
print(f"Hasil Desimal : {(kode_transaksi_2013) << 1}")<br>
Menggeser seluruh bit kode transaksi satu posisi ke kiri dan menampilkan hasilnya dalam bentuk biner dan desimal.
26. Menampilkan tanda selesai<br>
print("=== SELESAI ===")<br>
Menampilkan pesan bahwa program telah selesai dijalankan.
