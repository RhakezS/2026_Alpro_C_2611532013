## 2.1 Dasar Teori
Pemahaman teoritis mengenai struktur control flow merupakan syarat krusial sebelum melakukan perancangan sintaksis dan pengujian kode. Berikut adalah penjabaran konsep-konsep fundamental yang digunakan sebagai dasar dalam memecahkan studi kasus.

### 2.1.1 Konsep Dasar Perulangan
Perulangan merupakan salah satu struktur dasar dalam pemrograman yang digunakan untuk menjalankan suatu perintah secara berulang. Perulangan membantu programmer menghindari penulisan kode yang sama berkali-kali. Dalam Python, perulangan dapat digunakan untuk mengolah data, mencetak pola, dan menjalankan suatu proses dengan jumlah pengulangan tertentu. Python menyediakan beberapa jenis perulangan, salah satunya adalah for loop.

### 2.1.2 Perulangan For (For Loop)
For loop merupakan struktur perulangan pada Python yang digunakan untuk menjalankan suatu blok kode secara berulang berdasarkan data atau urutan yang diberikan. Setiap pengulangan akan mengambil satu nilai dari urutan tersebut dan menjalankan perintah yang berada di dalam blok for.

### 2.1.3 Fungsi Range (Range Function)
Fungsi range() digunakan untuk menghasilkan urutan bilangan yang dapat digunakan dalam perulangan. Fungsi ini sering digunakan bersama for loop untuk menentukan berapa kali suatu perintah akan dijalankan.

### 2.1.4 Perulangan Bersarang (Nested Loop)
Nested loop adalah perulangan yang terdapat di dalam perulangan lainnya. Pada nested loop, perulangan bagian luar disebut outer loop, sedangkan perulangan yang berada di dalamnya disebut inner loop. Outer loop biasanya digunakan untuk mengatur baris, sedangkan inner loop digunakan untuk mengatur isi atau kolom pada setiap baris.

### 2.1.5 Parameter End
Parameter end merupakan parameter pada fungsi print() yang digunakan untuk menentukan karakter yang ditampilkan setelah suatu nilai dicetak. Secara default, print() akan membuat baris baru setelah mencetak sebuah nilai. Dengan menggunakan end="", kursor tetap berada pada baris yang sama.

## 2.2 Kode Program

### 2.2.1 Program "perulangan_for1_2611532013"

```python
# Buat file dengan nama perulangan_for1_NIM.py
# Buat program untuk perulangan for dalam python
# Nama variabel ditambah 4 digit nim terakhir contoh: ulang_1234
# Program ini menggunakan fungsi input()

ulang_2013 = int(input("Masukkan jumlah perulangan: "))

for i_2013 in range(ulang_2013):
    print(f"Perulangan ke-{i_2013}")
```

**Penjelasan :**

1. Memasukkan nilai jumlah perulangan. Data yang dimasukkan diubah menjadi bilangan integer menggunakan int() dan disimpan pada variabel ulang_2013.
2. Program memulai mengeksekusi perulangan for menggunakan fungsi range(ulang_2013) untuk menghasilkan deret angka berurutan sebanyak nilai input tersebut.
3. i_2013 digunakan sebagai penunjuk yang menyimpan nilai angka pada setiap putaran, di mana perhitungannya dimulai dari angka 0.
4. Selama perulangan berjalan, program mengeksekusi instruksi cetak untuk menampilkan pesan "Perulangan ke-" yang diikuti oleh nilai urutan saat ini dari variabel i_2013.
5. Penggabungan teks string dan nilai variabel numerik dilakukan secara langsung menggunakan format f-string tanpa perlu mengubah tipe data angkanya secara manual.
6. Program akan terus mengulangi langkah tersebut secara berurutan hingga variabel i_2013 mencapai batas akhir dari deret angka yang ditentukan, lalu program berhenti.

### 2.2.2 Program "perulangan_for2_2611532013"

**Penjelasan :**

2. Program mencetak "Perulangan ke-0 sampai ke-" yang dipisahkan koma dengan operasi matematika ulang_2013 - 1 untuk menampilkan batas akhir.

5. Setelah siklus perulangan pertama benar-benar selesai, program mengeksekusi print() kosong yang berfungsi sebagai enter.
6. Program mencetak teks informasi kedua "Perulangan ke-1 sampai ke-".
7. Program menjalankan perulangan kedua dengan parameter rentang menjadi range(1, ulang_2013 + 1). Hal ini memaksa i_2013 memulai perhitungan berurutan dari angka 1.

### 2.2.3 Program "perulangan_for3_2611532013"

```python
# Buat file dengan nama perulangan_for3_NIM.py
# Buat program untuk perulangan for dalam Python
# Nama variabel ditambah 4 digit nim terakhir contoh: ulang_1234
# Program ini menggunakan fungsi input()

ulang_2013 = int(input("Masukkan jumlah perulangan: "))

jumlah_2013 = 0
for i_2013 in range(1, ulang_2013 + 1):
    print(i_2013, end=" ")
    jumlah_2013 = jumlah_2013 + i_2013

    if i_2013 < ulang_2013:
        print(" + ", end="")
    else:
        print(" = ", jumlah_2013, end="")
print()
print("Jumlah =", jumlah_2013)
```

**Penjelasan :**

1. Memasukkan nilai jumlah perulangan. Data yang dimasukkan diubah menjadi tipe integer menggunakan int() dan disimpan pada variabel ulang_2013.
2. Menginisialisasi jumlah_2013 dengan nilai awal 0.
3. Program memulai eksekusi perulangan for dengan fungsi range(1, ulang_2013 + 1). Variabel i_2013 akan otomatis berhitung mulai dari angka 1 hingga mencapai angka batas.
4. Program mencetak nilai i_2013 secara horizontal menggunakan end=" ". Setelah dicetak, nilai i_2013
langsung ditambahkan ke dalam jumlah_2013.
5. Program mengeksekusi percabangan if. Selama angka i_2013 masih lebih kecil dari batas akhir ulang_2013, program akan menyisipkan simbol tambah " + " di sebelah kanannya.
6. Apabila i_2013 sudah mencapai angka terakhir (kondisi else), program akan mencetak simbol " = ".
7. Setelah siklus perulangan selesai, program mencetak kesimpulan akhir berupa teks "Jumlah =" beserta nilai total di jumlah_2013.

### 2.2.4 Program "nested_for1_2611532013"

```python
# Buat file dengan nama nested_for1_NIM.py
# Buat program untuk perulangan for dalam Python
# Nama variabel ditambah 4 digit nim terakhir contoh: ulang_1234
# Program ini menggunakan fungsi input()

batas_2013 = int(input("Masukkan nilai batas: "))
for line_2013 in range(1, batas_2013 + 1):
    for j_2013 in range(1, (-1 * line_2013 + batas_2013 + 1)):
        print(".", end="")
    print(line_2013)
```

**Penjelasan :**

1. Memasukkan nilai batas maksimum. Data yang dimasukkan diubah menjadi tipe integer menggunakan int() dan disimpan pada variabel batas_2013.
2. Program mengeksekusi perulangan utama (outer loop) dengan fungsi range(1, batas_2013 + 1).
3. Di dalam setiap putaran baris, program mengeksekusi perulangan bersarang (nested loop) menggunakan variabel j_2013. Batas akhir perulangan menggunakan rumus matematika (-1 * line_2013 + batas_2013 + 1).
4. Rumus tersebut menyebabkan rentang perulangan semakin mengecil seiring bertambahnya angka baris. Selama perulangan bersarang ini berjalan, program mencetak karakter titik "." secara horizontal atau menyamping menggunakan end="".
5. Pemanggilan fungsi print(line_2013) tanpa parameter end="" otomatis berfungsi untuk mencetak angka sekaligus memindahkan posisi cetakan selanjutnya ke baris baru, sehingga membentuk pola titik-titik yang semakin menyusut dengan angka urutan di sisi kanannya.

### 2.2.5 Program "nested_for2_2611532013"

```python
# Buat file dengan nama nested_for2_NIM.py
# Buat program untuk perulangan for dalam Python
# Nama variabel ditambah 4 digit nim terakhir contoh: ulang_1234
# Program ini menggunakan fungsi input()

batas_2013 = int(input("Masukkan nilai batas: "))
for i_2013 in range(1, batas_2013 + 1):
    for j_2013 in range(1, batas_2013 + 1):
        print("*", end="")
    print() # pindah ke baris berikutnya
```

3. Di dalam setiap putaran baris, program menjalankan perulangan bersarang (nested loop) menggunakan variabel j_2013 dengan rentang ukuran yang sama persis seperti perulangan utamanya. Variabel ini pengendali jumlah karakter pada kolom horizontal.
4. Selama perulangan bersarang dieksekusi, program mencetak karakter bintang "*" secara terus-menerus ke arah samping kanan.
5. Perintah print() berfungsi layaknya enter di keyboard untuk memindahkan posisi pencetakan selanjutnya ke baris baru di bawahnya.
6. Proses ini diulang terus-menerus hingga seluruh putaran baris pada perulangan utama selesai dieksekusi, sehingga menghasilkan pola bangun persegi yang tersusun dari barisan karakter bintang.

### 2.2.6 Program "nested_for3_2611532013"

```python
# Buat file dengan nama nested_for3_NIM.py
# Buat program untuk perulangan for dalam Python
# Nama variabel ditambah 4 digit nim terakhir contoh: ulang_1234
# Program ini menggunakan fungsi input()

batas_2013 = int(input("Masukkan nilai batas: "))
for i_2013 in range(batas_2013+1):
    for j_2013 in range(batas_2013+1):
        print(i_2013 + j_2013, end=" ")
    print() # pindah ke baris berikutnya
```

4. Selama perulangan bersarang berjalan, program melakukan operasi matematika penjumlahan antara baris saat ini (i_2013) dengan kolom saat ini (j_2013). Hasil penjumlahan tersebut kemudian dicetak ke layar.
5. Pencetakan menggunakan parameter end=" " yang berfungsi menyisipkan spasi tepat di sebelah angka hasil penjumlahan dan deret angka terus tersusun secara horizontal.

### 2.2.7 Program "nested_for4_2611532013"

```python
# Buat file dengan nama nested_for4_NIM.py
# Buat program untuk perulangan for dalam Python
# Nama variabel ditambah 4 digit nim terakhir contoh: ulang_1234
# Program ini menggunakan fungsi input()

tinggi_2013 = int(input("Masukkan tinggi pola (bilangan genap, misal 10): "))

if tinggi_2013 % 2 != 0:
    print("Tinggi harus bilangan genap!")
else:
    a_2013 = tinggi_2013
    c_2013 = a_2013
    lebar_2013 = (2 * tinggi_2013) - 2

    for i_2013 in range(1, tinggi_2013 + 1):
        b_2013 = c_2013 + 1

        for j_2013 in range(1, lebar_2013 + 1):

            # Baris atas dan bawah
            if i_2013 == 1 or i_2013 == tinggi_2013:
                if j_2013 == 1 or j_2013 == lebar_2013:
                    print("#", end="")
                else:
                    print("=", end="")

            # Baris isi
            else:
                if j_2013 == 1 or j_2013 == lebar_2013:
                    print("|", end="")
                else:
                    if j_2013 == c_2013:
                        print("<", end="")
                    elif j_2013 == b_2013:
                        print(">", end="")
                    elif j_2013 == (lebar_2013 - c_2013):
                        print("<", end="")
                    elif j_2013 == (lebar_2013 - c_2013 + 1):
                        print(">", end="")
                    elif j_2013 > b_2013 and j_2013 < (lebar_2013 - c_2013):
                        print(".", end="")
                    else:
                        print(" ", end="")

        print()

        # Logika asli java
        a_2013 -= 2

        if a_2013 <= 0:
            c_2013 = (-a_2013 + 2)
        else:
            c_2013 = a_2013
```

**Penjelasan :**

1. Memasukkan angka tinggi pola. Program menggunakan operator modulus (% 2 != 0) untuk memvalidasi input, jika ganjil program mencetak peringatan dan berhenti, jika genap eksekusi berlanjut.
2. Program menginisialisasi variabel batas ukuran kolom horizontal pada lebar_2013, serta mengatur titik awal penempatan karakter dinamis menggunakan variabel a_2013 dan c_2013.
3. Program mengeksekusi perulangan utama i_2013 untuk mengendalikan pergerakan baris secara vertikal ke bawah, serta menetapkan nilai b_2013 sebagai batas untuk c_2013.
4. Di dalam setiap baris, program menjalankan perulangan bersarang j_2013 untuk mengevaluasi setiap titik koordinat dan mencetak karakter ke arah samping menggunakan parameter end="".
5. Percabangan pertama mengecek posisi baris. Jika merupakan baris teratas (1) atau terbawah (tinggi_2013), program mencetak bingkai horizontal pembatas berupa simbol # di sudut dan = di bagian tengah.
6. Di baris isi, program mencetak bingkai vertikal | di ujung kiri dan ujung kanan koordinat j_2013.
7. Di area dalam, program mengevaluasi deretan percabangan elif untuk membentuk dua pasang simbol < dan >. Area celah di antara sepasang simbol tersebut diisi dengan karakter titik ., sementara area luar diisi dengan spasi.
8. Setelah perulangan pencetakan kolom horizontal selesai, perintah print() dieksekusi untuk memindahkan pencetakan selanjutnya ke baris baru.
9. Di akhir setiap putaran baris, nilai a_2013 dikurangi 2. Jika nilainya mencapai titik tengah atau kurang dari 0 (<=0), rumus pada variabel c_2013 diubah untuk membalikkan arah, sehingga pola bentuk menyempit di atas kembali melebar di bagian bawah.

```python
print("=== PROGRAM JAM PASIR KRISTAL PALINDROMIK (PEKAN 5) ===")

n_2013 = int(input("Masukkan ukuran skala jam pasir (N): "))

# Border Atas:
print("#", end="")
for i_2013 in range(4 * n_2013 + 5):
    print("=", end="")
print("#")

# Jam Pasir Atas
for baris_2013 in range(n_2013, 0, -1):
    print("| ", end="")
    
    for spasi_2013 in range(2 * (n_2013 - baris_2013)):
        print(" ", end="")

    for angka_2013 in range(baris_2013, 0, -1):
        print(f"{angka_2013} ", end="")
        
    print("<*>", end="")

    for angka_2013 in range(1, baris_2013 + 1):
        print(f" {angka_2013}", end="")
        
    for spasi_2013 in range(2 * (n_2013 - baris_2013)):
        print(" ", end="")
        
    print(" |")

# Poros Titik Pusat Jam Pasir 
print("|", end="")
for spasi_2013 in range(2 * n_2013 + 1):
    print(" ", end="")
    
print("<*>", end="")

for spasi_2013 in range(2 * n_2013 + 1):
    print(" ", end="")
    
print("|")

# Jam Pasir Bawah
for baris_2013 in range(1, n_2013 + 1):
    print("| ", end="")
    
    for spasi_2013 in range(2 * (n_2013 - baris_2013)):
        print(" ", end="")

    for angka_2013 in range(baris_2013, 0, -1):
        print(f"{angka_2013} ", end="")    
        
    print("<*>", end="")

    for angka_2013 in range(1, baris_2013 + 1):
        print(f" {angka_2013}", end="") 
        
    for spasi_2013 in range(2 * (n_2013 - baris_2013)):
        print(" ", end="")
        
    print(" |")

# Border Bawah
print("#", end="")
for i_2013 in range(4 * n_2013 + 5):
    print("=", end="")
print("#")
```
