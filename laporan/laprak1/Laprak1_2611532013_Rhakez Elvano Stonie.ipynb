## 2.1 Dasar Teori
Dasar teori ini akan membahas tiga komponen utama yang digunakan dalam praktikum, yaitu Phyton sebagai bahasa pemrograman, VS Code sebagai teks editor untuk menyusun kode, serta GitHub sebagai platform untuk menyimpan dan mengelola repositori proyek.

### 2.1.1. Phyton
Python merupakan bahasa pemrograman tingkat tinggi yang memiliki sintaks sederhana dan mudah dipahami. Python menggunakan struktur kode yang relatif ringkas sehingga cocok digunakan untuk mempelajari dasar-dasar pemrograman. Python mendukung berbagai konsep pemrograman, seperti variabel, tipe data, operator, percabangan, perulangan, fungsi, dan struktur data. File program Python umumnya memiliki ekstensi .py. Program Python dapat dijalankan melalui interpreter Python yang akan menerjemahkan kode sehingga dapat dieksekusi oleh komputer.

### 2.1.2. Visual Studio Code
Visual Studio Code atau VS Code merupakan perangkat lunak code editor yang digunakan untuk membuat dan mengelola kode program. VS Code mendukung berbagai bahasa pemrograman, termasuk Python. VS Code menyediakan fitur seperti syntax highlighting, terminal, debugging, dan extension. Extension Python dapat digunakan untuk membantu proses penulisan dan menjalankan program Python. Dengan VS Code, programmer dapat membuat file, mengedit kode, menjalankan program, serta mengelola project dalam satu lingkungan kerja.

### 2.1.3. GitHub
GitHub merupakan platform berbasis internet yang digunakan untuk menyimpan dan mengelola repository Git. GitHub memungkinkan programmer menyimpan source code secara online sehingga project dapat diakses dan dikelola dengan lebih mudah. GitHub juga mendukung kolaborasi dalam pengembangan project melalui fitur seperti repository, commit, branch, dan pull request. Repository GitHub dapat digunakan sebagai tempat penyimpanan project Python yang dibuat menggunakan VS Code.

### 2.2.2 Membuat Repository GitHub

setelah program dibuat dan dilakukan uji coba pada VS Code, proyek dapat dihubungkan dengan GitHub. Repository digunakan untuk menyimpan kode program secara daring.

Tahapan yang dilakukan yaitu:

1. Membuat Repository baru pada Github.
2. Membuka folder proyek menggunakan VS Code.
3. Membuka terminal pada VS Code.
4. Menginisialisasi Git pada folder proyek.
5. Menambahkan file ke dalam staging.
6. Membuat commit.
7. Menghubungkan repository lokal dengan repository Github.
8. Melakukan push untuk mengunggah kode ke Github.

Contoh program yang digunakan adalah:

```python
git init
git add .
git commit -m "Initial commit"
git branch -M main
git remote add origin https://github.com/username/nama-repository.git
git push -u origin main
```

1. `git init`: Digunakan untuk membuat dan menginisialisasi repositori Git baru pada folder proyek lokal.
2. `git add .`: Digunakan untuk menambahkan seluruh berkas dan perubahan pada proyek ke dalam staging area.
3. `git commit`: Digunakan untuk menyimpan riwayat perubahan dari staging area disertai pesan deskriptif.
4. `git branch -M main`: Digunakan untuk mengubah atau mengonfirmasi nama cabang (branch) utama menjadi main.
5. `git remote add origin`: Digunakan untuk menghubungkan repositori lokal dengan repositori yang telah dibuat di GitHub.
6. `git push`: Digunakan untuk mengunggah berkas dan riwayat commit dari komputer lokal ke repositori GitHub.
