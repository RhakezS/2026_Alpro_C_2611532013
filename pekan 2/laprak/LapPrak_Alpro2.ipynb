## 2.1 Dasar Teori

Praktikum Pengenalan Python, VS Code, dan Koneksi ke GitHub membahas dasar penggunaan bahasa pemrograman Python, Visual Studio Code sebagai code editor, serta GitHub sebagai media penyimpanan dan pengelolaan kode program.

Praktikum Tipe Data dan Variabel membahas konsep dasar dalam bahasa pemrograman Python yang digunakan untuk menyimpan dan mengolah data. Tipe data menentukan jenis nilai yang digunakan dalam program, sedangkan variabel digunakan untuk menyimpan atau mereferensikan nilai tersebut.

1. Integer (int)<br>
Integer merupakan tipe data yang digunakan untuk menyimpan bilangan bulat, seperti 10, 25, dan -5.
2. Float (float)<br>
Float digunakan untuk menyimpan bilangan yang memiliki nilai desimal, seperti 3.14, 10.75, dan 2.5.
3. Complex (complex)<br>
Complex merupakan tipe data yang digunakan untuk menyimpan bilangan kompleks yang memiliki bagian real dan imajiner. Contohnya adalah 2 + 3j.
4. String (str)<br>
String merupakan tipe data yang digunakan untuk menyimpan teks atau kumpulan karakter. String ditulis menggunakan tanda kutip, contohnya "Python" dan "Algoritma".
5. Boolean (bool)<br>
Boolean merupakan tipe data yang memiliki dua nilai, yaitu True dan False. Tipe data ini biasanya digunakan untuk menyatakan kondisi benar atau salah.
6. Konstanta<br>
Konstanta merupakan nilai yang digunakan dalam program dan dianggap tetap selama program berjalan. Python tidak memiliki aturan khusus untuk membuat konstanta, tetapi biasanya nama konstanta ditulis menggunakan huruf kapital, seperti PI = 3.14.
7. Variabel<br>
Variabel digunakan sebagai tempat untuk menyimpan atau mereferensikan suatu nilai. Dalam Python, variabel dapat dibuat dengan memberikan nilai menggunakan operator =, misalnya nama = "Vano" atau umur = 18.
8. Input dan Output<br>
Python menyediakan fungsi input() untuk menerima data dari pengguna dan print() untuk menampilkan data ke layar. Data dari input() secara default bertipe str, sehingga perlu dilakukan konversi apabila data tersebut akan digunakan sebagai bilangan.
<br>
<br>
Dengan memahami tipe data dan variabel, mahasiswa dapat menentukan jenis data yang sesuai serta menggunakannya dalam proses pengolahan data pada program Python.

```python
#create a variable with integer value
a_2013 = 100
print("The type of variable having value",a_2013,"is",type(a_2013))

#create variable with float value
b_2013 = 10.2345
print("The type of variable having value",b_2013,"is",type(b_2013))

#create variable with complex value
c_2013 = 100 + 3j
print("The type of variable having value",c_2013,"is",type(c_2013))
```

### 2.2.3 Program Boolean_2611532013
</div>

```python
# Buat file dengan nama Boolean_2611532013
# Nama variabel ditambah 4 digit nim terakhir contoh: nilai_1234
# Deklarasi variabel dengan tipe data Boolean
is_lulus_2013 = True
is_cumlaude_2013 = True

# Menggunakan Boolean
nilai_2013 = 85
batas_lulus_2013 = 75

# Menantukan nilai Boolean dari kondisi
status_kelulusan_2013 = nilai_2013 >= batas_lulus_2013 # Hasilnya akan True

print("=== Check Kelulusan ===")
print("Nilai: ",nilai_2013)
print("Apakah lulus?: ",status_kelulusan_2013)
if is_lulus_2013 and is_cumlaude_2013:
    print("Selamat, Anda lulus dengan predikat Cum Laude!")
```

6. Deklarasi status_kelulusan_2013 = nilai_2013 >= batas_lulus_2013 sebagai boolean

### 2.2.5 Program Konstanta_2611532013
</div>

```python
# Buat file dengan nama Konstanta_2611532013
# Program ini menggunakan konstanta untuk menghitun luas lingkaran
# nama variabel ditambah 4 digit nim terakhir contoh: jari_2013

from typing import final
PI: final = 3.14
print("pi: %f" % (PI))
jari_2013 = float (input('Masukkan nilai jari-jari:'))
luas_2013 = PI * jari_2013 * jari_2013
print("Luas lingkaran dengan jari-jari %.2f adalah %.2f" % (jari_2013, luas_2013))
```

<div style="font-family: 'Times New Roman', Times, serif;font-size: 18px;">

Penjelasan:
1. Pemanggilan Final dari library typing from typing import Final
2. Deklarasi PI: Final = 3.14 sebagai konstanta
3. Cetak ("Pi: %f" % (PI))
4. Nilai jari_2013 = float(input("Masukkan nilai jari-jari: "))
5. Nilai luas_2013 = PI * jari_2013
6. Cetak ("Luas lingkaran dengan jari-jari %.2f adalah %.2f" % (jari_2013,luas_2013))
